## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
#Question: Given a page's search and engagement signals over the last 90 days, which pages should a content reviewer look at first for a possible refresh?

Unit of analysis: one row = one content page belonging to one client (not a client-level or day-level unit).

Decision improved: which pages get prioritized for manual review/refresh out of a large backlog, given limited reviewer time.

Who acts on it: a content strategist or SEO reviewer who can only check a fixed number of pages per cycle (e.g. top 50).

Cost of a wrong call: two-sided. Flagging a page that isn't actually a problem wastes a limited review slot that could've gone to a real opportunity. Missing a genuinely declining, high-traffic page means it keeps losing visibility and clicks silently until the next review cycle.

Why data/ML helps here, not just a model for its own sake: in this slice, 43.8% of pages already show some decline signal — far more than any reviewer can check. A hand-written rule only gets ~12 of its top 50 picks right (Precision@50 = 0.24); a learned ranking gets ~37 of 50 right (0.74). That's a 3x improvement in how well limited review time is spent — a concrete decision improvement, not complexity for its own sake.
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/maha-naveed77/Flyrank-Internship"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
else:
    while not os.path.isdir("data/raw") and os.getcwd() != "/":
        os.chdir("..")

print("Working dir:", os.getcwd())
assert os.path.exists("data/raw/content_refresh_anonymized.csv"), "starter CSV not found — are you at the repo root?"

import pandas as pd

df = pd.read_csv('data/raw/content_refresh_anonymized.csv')
filt = df[(df['impressions_90d'] > 0) & (df['content_age_days'] >= 90)]

print("Rows after starter filter:", len(filt))
print(filt['trend_direction'].value_counts())

pct_down = (filt['trend_direction'] == 'down').mean()
declining_with_demand = filt[(filt['trend_direction']=='down') & (filt['impressions_90d']>=100)]

print(f"Share currently trending down: {pct_down:.1%}")
print(f"'declining_with_demand' pages: {len(declining_with_demand)} ({len(declining_with_demand)/len(filt):.1%})")
print("Median days since last update:", filt['days_since_last_update'].median())
print("Distinct clients in this slice:", filt['client_id'].nunique())
```

```python
#Can say: these pages showed a declining trend or weak engagement pattern over the observed 90-day window (observed). The ranked list is decision-support — it tells a reviewer where to look first, not what to change. Any link between a signal (low CTR, thin content) and decline is directional — associated with, not proven to cause.

#Can't say: that refreshing a flagged page will cause recovery — that needs a causal experiment this data doesn't provide. That I've predicted or reverse-engineered Google's ranking algorithm — I'm only using FlyRank's own observed signals. That trend_direction == "down" is a true future outcome — it's a proxy from the current window, and I'll likely replace it with a real future-window label in a later week.
```
