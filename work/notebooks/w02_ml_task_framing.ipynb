## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# Precision@50. A reviewer only checks a fixed number of pages per cycle, so what matters is how many of the top 50 are genuinely worth reviewing — not overall accuracy across all 30,000 rows. Baseline rule: 0.240 (~12/50 right). Random forest: 0.740 (~37/50 right).
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
#The fixed baseline rule applies the same fixed weights to every page regardless of how signals actually combine. The Precision@50 gap (0.240 vs. 0.740) shows those combinations matter — the model learns which signal interactions predict decline, something a linear weighted rule can't capture. This is backed by numbers, not asserted.
```
